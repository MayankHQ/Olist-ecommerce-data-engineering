```python
# Read raw orders
orders_df = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(
        "abfss://ecommerce@mayankecommerce017.dfs.core.windows.net/raw/orders/olist_orders_dataset.csv"
    )
)

# Bronze destination
bronze_path = "abfss://ecommerce@mayankecommerce017.dfs.core.windows.net/bronze/orders"

# Write as Delta
orders_df.write \
    .format("delta") \
    .mode("overwrite") \
    .save(bronze_path)
```
