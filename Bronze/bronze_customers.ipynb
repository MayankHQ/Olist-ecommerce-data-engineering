```python
customers_df = (
    spark.read
    .options(header = "true", inferSchema = "true")
    .csv(
        "abfss://ecommerce@mayankecommerce017.dfs.core.windows.net/raw/customers/olist_customers_dataset.csv"
    )
)

bronze_path = "abfss://ecommerce@mayankecommerce017.dfs.core.windows.net/bronze/customers"

customers_df.write \
    .format("delta") \
    .mode("overwrite") \
    .save(bronze_path)
```

```python
bronze_customers = spark.read.format("delta").load(bronze_path)

print("Raw rows:", customers_df.count())
print("Bronze rows:", bronze_customers.count())

bronze_customers.show(5)
```
